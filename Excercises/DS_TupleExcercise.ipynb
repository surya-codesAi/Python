```python
#Immutable: An immutable object cannot be modified after it is created. 
# However, it can be used to create a new object of another type.
t=(10,20,40,50)
print(type(t))
l=list(t)
print(l, type(l))
print(t, type(t))#see here original tuple remaind unchanged 
#after conversion to list can perform all list operations
l.pop()
print(l)
l.append(500)
print(l)
#again can convert it to tuple
t=tuple(l)
print(t)
```

```python
#A tuple is called ordered because Python preserves the position of each element inside the tuple.
t=(10,20,40,50)
print(t[0])
print(t[2])
```

```python
#slicing
t=("surya","krishna",10,24.56,"ram",100,57,False)
print(t[:3])
print(t[2:])
print(t[3:6])
print(t[::1])
print(t[::-1])
print(t[::2])
```

```python
#tuple concatenating
#The + operator joins the two tuples in their existing order.
t1=(10,20,30)
t2=(40,50,60)
t=t1+t2
print(t)#It does not modify tuple1 or tuple2.It creates a new tuple and stores it in tuple3.
#+ means join/concatenate, not mathematical addition.for list and tuple
```
