```python
#re is Python's built-in module for working with regular expressions (regex),
#  which are patterns used to search, match, extract, or replace text.
#A module is simply a Python file containing code that you can reuse in another Python program.
#Python also comes with many additional tools, but they aren't automatically loaded.
#thats why we use import
#Because Python has thousands of pieces of functionality.
#  Loading everything automatically would be unnecessary.
#normally in python when backslash is used like \n means Make a new line.
#generally using \ has some special meaning in python like new line while in regex
#we use r means raw string because dont consider \ inside the pattern any meaning
print("Hello\nWorld")#see it has the meaning 
print(r"Hello\nWorld")#here no meaning considering it as a just a part of a str value
#A raw string is a string prefixed with r that tells Python
#  to treat backslashes mostly as literal characters.
```

```python
import re
#DIGIT
#\d matches a digit from 0 to 9.
text="i am 25 years old boy"
print(re.findall(r"\d",text))#\d finds each digit separately
#+ — One or More
print(re.findall(r"\d+",text))#groups of one or more digits
```

```python
#\w — Word Character  matches letters, digits, and underscore.
t="hare krishna_108rounds * % $"
print(re.findall(r"\w",t))
print(re.findall(r"\w+",t))
```

```python
#\s — Whitespace
t="hare krishna_108rounds * % $"
print(re.findall(r"\s",t))
```

```python
#Notice the capital letter reverses the meaning. \D, \W, \S
text="i am 25 years old boy% # $"
print(re.findall(r"\d",text))
print(re.findall(r"\D",text))
print(re.findall(r"\w",text))
print(re.findall(r"\W",text))
print(re.findall(r"\s",text))
print(re.findall(r"\S",text))
```

```python
#same with grps
text="i am 25 years old boy% # $"
print(re.findall(r"\d+",text))
print(re.findall(r"\D+",text))
print(re.findall(r"\w+",text))
print(re.findall(r"\W+",text))
print(re.findall(r"\s+",text))
print(re.findall(r"\S+",text))
```

```python
#. is a wildcard that matches almost any single character.
t="hare krishna hare rama haare krishna"
print(re.findall(r"h.r",t))
print(re.findall(r"h..r",t))
```

```python
#Character Sets [] Square brackets let you specify allowed characters.
t="hare krishna hare rama haare krishna"
print(re.findall(r"[kha]",t))##Match one character that is either k, h, or a.
```

```python
#Character ranges Instead of writing: [abcdefghijklmnopqrstuvwxyz] you can write:[a-z]
#Meaning:Any lowercase letter from a to z.
#Similarly:[A-Z] uppercase letters. [0-9] for digits.So:[a-z] [A-Z] [0-9] are very common patterns.
t="hare krishna hare rama haare krishna"
print(re.findall(r"[a-k]",t))
print(re.findall(r"[A-Z]",t))
#Combining ranges You can combine them: [a-zA-Z0-9] means:lowercase letters OR uppercase letters OR digits.
```

```python
#[^...] — NOT these characters Inside square brackets, ^ at the beginning means not.
#[^0-9] means:Any character that is NOT a digit.
t="krishna_108"
print(re.findall(r"[^0-9]",t))
```

```python
#re.search() searches the entire string and returns the first matching occurrence.
t = "My age is 25 from 2000"
result = re.search(r"\d+", t)
print(result.group())
#If there is no match: ans is None
m = "My age is "
result = re.search(r"\d+", m)
print(result.group())
```

```python
#re.match() checks whether the pattern matches at the beginning of the string.
t = "123 Hello"
result = re.match(r"\d+", t)
print(result.group())
t = "Hello 123"
result = re.match(r"\d+", t)
print(result)
```

```python
#re.fullmatch() succeeds only when the entire string matches the pattern.
text = "12345"
result = re.fullmatch(r"\d+", text)
print(result.group())#here it works
text = "12345abc" #but not here

r=re.fullmatch(r"\d+", text)#returns None.This is useful for validation.
print(r)
```

```python
#re.findall() returns all matching occurrences as a list.
text = "I have 10 apples and 20 oranges"
numbers = re.findall(r"\d+", text)
print(numbers)
#re.sub() replaces text matching a pattern with another string.
text = "I have 123 apples"
result = re.sub(r"\d+", "many", text)#sub → substitute → replace
print(result)
#re.split() splits a string wherever the regex pattern matches.
text = "apple,banana mango"
result = re.split(r"[, ]", text)
print(result)
```

```python
#*, +, ? These three are called quantifiers.
#A quantifier tells regex how many times the previous pattern can occur.
#a as our basic pattern.
# + → one or more  ,a+  One or more a's.
#* → zero or more   a*  Zero or more as.
#main differnce + → at least 1  ,* → can be 0
#zero or only one   a?  a is optional.
#{n} exactly n numbers,{n,} atleat n numbers and max can be any,{n,m} means min and max value
#start() returns the starting index of the match.
#end() returns the index immediately after the match.indexing starts by zero
#^ and $ These are positions, not characters.^ Means:The beginning of the string
#$ Means:The end of the string eg: r"^Hello" and r"Krishna$"
#why end() produces next index after match
#  The ending index tells you where the match stops, and that position itself is NOT part of the match.
#like start index is included but end index not 
```

```python
#finditer() also finds all matches, just like findall().
#findall() gives you the actual text gives output in form of list
#finditer() gives you information about each match object
t="i am 25 years old born in year 2000 at Jan31st"
print(re.match(r"\d+",t))#because at very begining there is no match
tm="25 years old born in year 2000 at Jan31st"
print(re.match(r"\d+",tm).group())
#we always need to use group() for a match objects
print(re.search(r"\d+",t))
print(re.search(r"\d+",t).group())#will produce only 1st occurance
print(re.findall(r"\d+",t))#returns all occurances
m=re.finditer(r"\d+",t)#this also produces like findall but also additional informations
for i in m:
    print(i.group())
    print(i.start())
    print(i.end())
```

```python
#A boundary is simply:The place where a word character and a non-word character meet.
#A-Z a-z 0-9 _  are word characters.Everything else, such as space . , ! ? is a non-word character.
```

```python
#re.compile() converts a regex pattern into a compiled regex object 
# that you can reuse for multiple searches.
pattern = re.compile(r"\d+")
t1="i am 25 years old"
t2="my brother 21 years old"
print(pattern.findall(t1))
print(pattern.findall(t2))#can reuse it any number of times
```

```python
#  . means any single character inbetween but \. means actually a . character is present in str
t="my email id is surya@gmail.com"
print(re.findall(r"\w+\.\w+",t))#beacuse actual . is present inside 
```
