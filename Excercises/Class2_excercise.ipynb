```python
#positive indexing in String
name="Krishna"
print(name[0])
print(name[1])
print(name[2])
print(len(name))
print(name[6])
#print(name[7]) error comes index out of range because index always starts from 0 to length-1.
```

```python
#negative indexing in String
name="krishna"
print(name[len(name) - 1])
print(name[-1])
print(name[-2])
print(name[-3])
```

```python
#String slicing
#Slicing means taking  a part of a string. [start:stop] start is included stop is not included
name="radha krishna"
print(name[0:5])
print(name[5:])
print(name[:])#Prints the entire string, from beginning to end.
print(name[:5])#Prints characters from the beginning up to, but not including, index 5.
print(name[6:])#Prints characters starting from index 6 until the end.
```

```python
#in slicing when start staring from zero move to particular index which is in left to right direction 
#if we give step as -1 that is right to left will not work
var="radhakrishna"
print(var[0:5:-1])#no error answer will be empty 
#here because Start tells me where I begin. Stop tells me where I finish. Step tells me which direction I walk.
# step says move right to left.So Python starts at index 0 and tries to move left:
#But there is nothing to the left of index 0.
print(var[::-1])#it will work because we think that if start is empty it will take zero not at all
#it chooses by auto the appropriate starting point based on the direction of the step
#If start and stop are empty, the step decides where to start and where to stop.
```

```python
#name[start:stop:step] slicing with a step 
#The step tells Python how many positions to move each time.
#  It does not mean "print every nth character."
name="radha krishna"
print(name[0::1])#So Python moves one position at a time
print(name[::])#Python's slicing rule is that to take step 1 as default value.
print(name[::2])
print(name[0:5:1])
print(name[5::1])
print(name[6::2])
#reverse using steps
print(name[::-1])#start = omitted stop  = omitted step  = -1
#Because the step is negative, Python automatically chooses the last position as the starting point.
#When start is omitted, Python chooses the appropriate end based on the direction of step.
print(name[-1::-1])
print(name[-1:-8:-1])
```

```python
#String methods
s="Krishna asks Ram to come vraj by today evening"
print(s.upper())#converts all to upper case
print(s.lower())#converts all to lower case
print(s.swapcase())#it will swap the cases vice versa
print(len(s))
print(s.find("j"))#Returns the index of the first occurrence
#If the substring is not found, it returns -1.
print(s.capitalize())#converts the first chr of  string to upcase and converts the remaining chrsto lowcase
print(s.title())#converts the first chr of each word to upcase and the remaining chr of each word to lowcase
print(s.index("j"))#returns the index of the first occurrence of the specified substring
#If the substring is not found, index() raises a ValueError
print(s.count("a"))
print(s.replace("a","$"))
print(s.strip())#It removes spaces from both sides, not from between words
r="  hi krishna"
print(r)
print(r.lstrip())
m="hi ram   "
print(m.rstrip())
print(s.split())#divides a string into a list of substrings based on a specified separator
#boolean checks
print(s.startswith("K"))
print(s.endswith("k"))
print(s.isalnum())
print(s.isalpha())
print(s.isnumeric())
```
