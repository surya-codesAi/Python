```python
s = {1, 2, 3}
s.update([10, 20])
print(s)
s.update((10, 20))
print(s)
#Because set.update() doesn't specifically require a list or a tuple.
#It expects an iterable — something Python can go through one element at a time.
```

```python
#A set is unordered because a set does not store elements according to a fixed position or index.
#Python's set uses a hash table internally. The position where an element is stored depends on its hash value,
#  rather than the order in which you inserted it.
s={10,20,30,40,50}
print(s, type(s))#see insertion order is not followed
#so indexing not possible
```

```python
#hash table is a data structure that stores values in locations determined using their hash values.
#A hash value is a number generated from an object by a hashing function.
#Python uses that number to help decide where to store/find the value.
a="krishna"
print(hash(a))#hash value
#Set insertion: When adding an element to a set, Python uses the element's hash value 
# to locate its possible position and equality comparison (==) to determine
#  whether an equal element already exists.
```

```python
#Union | → combine everything means: take all elements from both sets, without duplicates.
s1={1,2,3,4,5}
s2={4,5,6,7,8,9}
us=s1|s2
print(us)
print(s1.union(s2))#this also correct
#Intersection & → common elements means: elements that are present in both sets
ins=s1&s2
print(ins)
print(s1.intersection(s2))
#Difference / Minus - → elements only in the first set,elements that are in the first set but not in the second set.
difs=s1-s2
print(difs)
print(s2.difference(s1))
#Symmetric Difference ^ Returns elements that are in either set, but not in both.
sym=s1^s2
print(sym)
print(s1.symmetric_difference(s2))
```
