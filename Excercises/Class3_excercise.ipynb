```python
a = 10
b = 20
operator = input("Enter any operators")
if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    print(a / b)
else:
    print("Invalid operator")
```

```python
# signal - red, yellow, green
signal = input("Enter a signal ")
if signal == "red":
    print("Stop")
elif signal == "yellow":
    print("Get ready")
elif signal == "green":
    print("Go")
else:
    print("Invalid signal")
```
