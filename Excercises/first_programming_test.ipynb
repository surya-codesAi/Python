```python
current_units=int(input("enter the units"))
if current_units<=100:
    charge=current_units*2
    print(f"units: {current_units} Total Electricity Bill: Rs.{charge}")
elif current_units>100 and current_units<=200:
    charge=current_units*3
    print(f"units: {current_units} Total Electricity Bill: Rs.{charge}")
elif current_units>200:
    charge=current_units*5
    print(f"units: {current_units} Total Electricity Bill: Rs.{charge}")
else:
    print("no bill generated for ranges")
```

```python
initial_balance=10000
withdrawal=int(input("enter the withdrawal amount"))
if withdrawal>0:
    if withdrawal<initial_balance:
        deduction=initial_balance-withdrawal
        print(f"balance amount: {deduction}")
    else:
        print(" insufficient balance")
else:
    print("withdrawal is not possible")
```

```python
numbers = [10, 20, 30, 40, 50]
numbers[0]=100
print(f"after changing the first element: {numbers}")
numbers.insert(2,25)
print(f"after inserting element in 2nd index: {numbers}")
numbers.append(60)
print(f"after adding the  element: {numbers}")
numbers.remove(40)
print(f"after removing a  element: {numbers}")
numbers.extend([70,80])
print(f"after extending the elements: {numbers}")
```

```python
marks =[85, 60, 95, 70, 80]
lis1=sorted(marks)
print(f"after sorting the elements in ascending order: {lis1}")
lis2=sorted(marks,reverse=True)
print(f"after sorting the elements in descending order: {lis2}")
print(f"max element in a list: {max(marks)}")
print(f"min element in a list: {min(marks)}")
print(f"sum of element in a list: {sum(marks)}")
avg=sum(marks)/len(marks)
print(f"avg of element in a list: {avg}")
```
