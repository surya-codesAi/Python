```python
#A for loop is used to repeatedly execute a block of code for each item in an iterable,
#like list, tuple, string, set,dictionary.Take each item from the collection, one by one, and do something with it.
#for variable in iterable:
    # code to execute
n=[10,20,30,40,50,60]
for num in n:#takes one item at a time from the list and puts it into num.
    print(num)
for i in range(1,11):
    print(i)

#range() is a built-in Python function used to generate a sequence of integers, 
# commonly for controlling iterations in a for loop.
```

```python
#print() displays output and, by default, ends with a newline (\n).
#end specifies what print() should place after the output. Its default value is "\n".
#print() automatically adds a new line
name="hare krishna"
for i in name:
    print(i)
```

```python
#A while loop repeatedly executes a block of code as long as a given condition is True.
#while condition:
    # code to execute
#differnce is that for is often used when number of iterations is known
#while often used when number of iterations depends on a condition
#Initialization → give the starting value
#Condition → decide whether the loop should continue
#Update → change the value so the loop can eventually stop

#A typical while loop consists of initialization, a condition, and an update.
#  The update should eventually make the condition False, allowing the loop to terminate.
#For a for loop, you usually don't need to manually write initialization and update.
i=5
while i>=0:
    print("krishna")
    i-=1
```

```python
n="hare krishna hare rama"
for i in n.split():
    print(i)
```

```python
n=["krishna","radha","rama","sita"]
for i in n:
    print(list(i.upper()))
output=[]
for i in n:
    i.upper()
    output.append(i)
    print(output)
print(output)
```

```python
e={"id":[10,20,30,40],
   "name":["surya","krishna","ram","sita"],
   "place":["salem","vraj","ayodhya","janaka"]
   }
for i in e:
    print(i)#Python's default behavior for a dictionary is:Iterate over the dictionary's keys.
for i in e.keys():
    print(i)
for i in e.values():
    print(i)
for i in e.items():
    print(i)#Why tuples?Because each item represents two things together:
#items() returns an iterable of key-value pairs, where each pair is represented as a tuple (key, value).
```

```python
#break is a statement that immediately stops the loop and moves execution to the code after the loop.
#once it finds break ,the entire loop stops.comes out of the loop
#continue skips the code in the current iteration and moves to the next iteration of the loop.
#The continue immediately jumps back to the while condition.
```

```python
#enumerate() is a built-in Python function that lets you loop through 
# an iterable while getting both the index (position) and the value at the same time.
#enumerate() → index + value
#Used mainly with lists, tuples, strings, etc.
names=["surya","krishna","ram","sita","radha"]
for i,v in enumerate(names):
    print(i,v)
```
