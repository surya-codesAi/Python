```python
#1.Tuple details
#Create a tuple containing the names of five cities.
t=("salem","banglore","chennai","mumbai","kochi")
print(f"five cities: {t}")
print(f"the city at index 2: {t[2]}")
print(f"last three cities: {t[-3:]}")
print(f"city using count: {t.count("kochi")}")
print(f"index of a selected city: {t.index("mumbai")}")
```

```python
#2. tuple to list conversion
t=(10,20,30,40,50)
print(type(t))
lis=list(t)
print(f"Convert the tuple into a list: {lis}")
lis[2]=35
print(f"Replace 30 with 35: {lis} ")
lis.append(60)
print(f"Add 60 to the list: {lis}")
lis.remove(10)
print(f"Remove the first element: {lis}")
#Convert the updated list back into a tuple
t=tuple(lis)
print(f"final tuple after conversion: {t}")
```

```python
#3. combine and process tuples
colors1 = ("red", "blue", "green")
colors2 = ("yellow", "blue", "white")
colors=colors1+colors2
print(f"Print the combined tuple: {colors}")
print(f"Count how many times blue appears: {colors.count("blue")}")
print(f"alternate elements of combied tuple: {colors[::2]}")

```

```python
#4.find common and unique elements
morning_batch = {"Arun", "Bala", "Charan", "Divya"}
evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}

print(f"students present in either batch: {morning_batch|evening_batch}")
print(f"students present in both batches: {morning_batch&evening_batch}")
print(f"students present only in the morning batch: {morning_batch-evening_batch}")
print(f"students present only in the evening batch: {evening_batch-morning_batch}")

```

```python
#5.remove duplicate names
names = ["Ravi", "Priya", "Ravi", "Kumar", "Priya", "Anu", "Kumar"]
print(f"initial list with duplicates: {names}")
s=set(names)
print(f"Convert the list into a set: {s}")
print(f"convert the set back into a list: {list(s)}")
```

```python
#6 update product sets
available_products = {"Laptop", "Mouse", "Keyboard"}
new_products = {"Monitor", "Mouse", "Printer"}
available_products.add("Webcam")
print(f"Add Webcam to available_products: {available_products}")
available_products.remove("Keyboard")
print(f"Remove Keyboard from available_products: {available_products}")
print(f"Combine using the union operator: {available_products|new_products}")
print(f"Find the products common to both sets: {available_products&new_products}")
print(f"Find the products available only in available_products: {available_products-new_products}")

```

```python
#7: STUDENT MARKS DICTIONARY
marks = {"Arun": 85, "Bala": 92, "Charan": 78, "Divya": 88}

print(f"marks of Bala: {marks["Bala"]}")
print(f"marks of Charan using get(): {marks.get("Charan")}")
print(f"Display all student names: {marks.keys()}")
print(f"Display all marks: {marks.values()}")
print(f"Display all student names and marks: {marks.items()}")
```

```python
#8: DEPARTMENT STAFF DICTIONARY

department_staff = {
    "HR": ["Ravi", "Sita"],
    "IT": ["Varun", "Ananya"],
    "Sales": ["Kiran", "Meena"]
}

print(f"Print the complete dictionary: {department_staff}")
print(f"all employees in the IT department: {department_staff["IT"]}")
print(f"first employee in the HR department: {department_staff['HR'][0]}")
print(f"last employee in the Sales department: {department_staff['Sales'][-1]}")
department_staff["HR"].append("Ram")
print(f"Add one more employee to the HR: {department_staff}")

```

```python
#9: MANAGE A SHOPPING CART DICTIONARY

cart = {"Pen": 10, "Notebook": 50, "Bag": 500}
cart.update({"Pencil":5})
print(f"Add pencil with a value of 5: {cart}")
cart.update({"Notebook":60})
print(f"Update the value of Notebook to 60: {cart}")
print(f"Retrieve the value of Bag: {cart.get("Bag")}")
print(f"remove pen:the return value is  {cart.pop("Pen")}")
print(f"Display the remaining dictionary: {cart}")
cart.clear()
print(f"after clear: {cart}")

```

```python
#10: CREATE A SUBJECT MARKS DICTIONARY USING ZIP

subjects = ["Tamil", "English", "Math", "Science"]
marks = [90, 85, 95, 88]
d=dict(zip(subjects,marks))
print(f"Create a dictionary using zip: {dict(zip(subjects,marks))}")
print(f"Display the dictionary: {d}")
print(f"add Social with a mark of 80: {d.setdefault("Social",80)}")
print(d)
print(f"existing key Math with a mark of 100: {d.setdefault("Math",100)}")
print(d)
```
