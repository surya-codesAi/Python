```python
#DICT are orderd,dictionaries maintain the insertion order of their key-value pairs.
dic={"name":"surya","age":25,"place":"salem"}
print(dic)#follows insertion order
#dictionary is mutable because we can change its contents after creating it.
dic["age"]=24
dic["country"]="india"
print(dic)
#dictionary does not allow duplicate keys because a key is used to uniquely identify a value.
dic={"name":"surya","name":"krishna"}
print(dic)#doesn't keep two name keys. The last value replaces the previous value
#dictionaries always enforces the idea:One key → one value
#it uses the hash of 101 to find where that key belongs.If you later write:
#dic["name"] = "krishna"
#Python finds that name already exists and updates its value instead of creating another name.
```

```python
#Duplicate keys are not allowed, but duplicate values are perfectly fine
dic={"name":"surya","last_name":"surya","age":25,"place":"salem"}
print(dic)#allows duplicate values 
#dictionaries are designed so that assigning a value to an existing key means
#  “update this key”, not “create a duplicate key.”
#doesnt throws error because dictionaries are mutable and are commonly used for updating information.
dic["last_name"]="krishna"#for this reason
print(dic)
print(type(dic))
```

```python
#operations
dic={"name":"surya","last_name":"surya","age":25,"place":"salem"}
print(dic.keys())#return all keys
#why dict_keys() in output because keys() does not return a list.
# It returns a special object called a dictionary view.
print(dic.values())
print(dic.items())#returns a dictionary view containing key-value pairs, 
#where each pair is represented as a tuple.not as list of tuples
print(dic.get("place"))#advantage of get() is that it doesn't give an error if the key doesn't exist.
print(dic["place"])
#to add new or modifying already existing
dic.update({"place":"vraj"})#modifying
dic.update({"city":"mathura"})#adding new
print(dic)
age=dic.pop("age")#Removes a specified key and returns its value.
print(dic)
print(age)
dic.popitem()#Removes and returns the last inserted key-value pair.
print(dic)
dic.setdefault("pin",108)#If the key doesn't exist, it creates the key with a default value.
print(dic)
```

```python
#shallow copy
#A shallow copy creates a new outer object, but the objects contained 
# inside it are not copied; instead, their references are shared.
d = {
    "name": "krishna",
    "marks": [80, 90]
}#here outer obj is dict but list inside it is shared it is nested object
#as nested obj is shared making changes in it changes the original also
d1=d.copy()
print(d)
print(d1)
print("-"*10)
#now will change name it will not affect the original
d1["name"]="ram"
print(d)
print(d1)
print("-"*10)
#now will change the list which is shared among two will affect the original 
d1["marks"].append(100)
print(d)
print(d1)
#A deep copy creates a completely independent copy of an object, including its nested objects.
#so oringial will be unaffected in anyway
#d1 = copy.deepcopy(d)
```

```python
#zip() is a Python built-in function that pairs elements from multiple iterables position by position.
names = ["John", "Alex", "Sam"]
marks = [80, 90, 75]

d = dict(zip(names, marks))

print(d)#Pair the elements, then convert the pairs into a dictionary.
```
