```python
#A list is a collection of multiple values stored in a single variable.
l1=[10,20,40,60]
print(type(l1))
l2=["krish","surya","ram","vasanth"]
print(type(l2))
l3=[10,"ram",10.18,"krishna",100]#differnt datatypes
print(type(l3))
l4=[10,"ram",10.18,"krishna","ram",100]#duplicate values are allowed
print(l4)
```

```python
#accessing the list elements
#Lists use index numbers, starting from 0.
l=["ram","krish","surya",60,10.19]
print(len(l))
print(l[1])
#print(l[5])#list index out of range as index starts from 0
print(l[4])
print(l[-1])
print(l[-2])
```

```python
#Lists are mutable, meaning we can change their values means updating the values
#list methods that modify the list usually change the original list directly.
l=["ram","krish","surya",60,10.19]
l[2]="shiv"
print(l)
l[2]="shyam"
print(l)
l.append("meer")#Adds one item at the end.
print(l)
l.append("rith")
print(l)
l.extend([87,99])#Adds multiple items to the end.expects an iterable element as a single argument
print(l)
l.append(["red","blue"])#here still it is one accepts one argument here list is a single argument
print(l)
l.insert(2,"radha")#Adds an item at a specific index.
print(l)
l.remove("rith")#Removes a specific value.
print(l)
l.pop(8)#Removes an item using its index.without index removes last item
print(l)
#l.clear() Removes everything from the list.returns empty list
print(l.index("krish"))#Finds the index of a value.
l.append("ram")
print(l.count("ram"))#how many times a value occurs.
```

```python
#sorting
l=["ram","krish","surya",60,10.19]
#l.sort()  error:not supported between instances of 'int' and 'str'
l1=[100,96,108,77,94]
l1.sort()#by default Sorts the list in ascending order.
print(l1)
l1.sort(reverse=True)#for descending order
print(l1)
l.reverse()#simply reverses the existing order.
print(l)
l2=l.copy()#makes a copy and stores in new list
print(l2)
```

```python
#commonly used funcs with list
l=["ram","krish","surya",60,10.19]
print(len(l))
l1=[100,96,108,77,94]
print(max(l1))
print(min(l1))
print(sum(l1))
#sorted() creates/returns a sorted result and leaves the original list unchanged.
#.sort() changes the original list.
lis=[71,77,89,75,63]
lis1=sorted(lis)
print(lis1)
print(lis)
lis2=sorted(lis,reverse=True)
print(lis2)
print(lis)
```

```python
#In Python, some values are treated as False when Python checks them in a condition.
#Common falsy values:0,"", None, [], (), False
#Everything else is generally considered truthy.
l= [0, None, False, "", 10]
print(all(l))##returns True only when every item is truthy.it is like AND
print(any(l))#returns True if at least one item is truthy.it is like OR
#An empty list [] is falsy, but any non-empty list is truthy — even if it contains None, False, or 0.
```

```python
l=[10,50,30,70,20,60]
print(l[-2:2])#empty because by default step is 1 moves from left to right so after -2 ,-1 is there
print(l[-2:2:-1])#now it will work because it will go right to left
l.extend([150,200])#here why we give as list because always multiple elements means collection of elemts
print(l)
```
